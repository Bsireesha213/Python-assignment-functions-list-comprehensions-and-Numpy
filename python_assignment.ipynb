```python
#format_name(first,last="") that returns "last,first" if last name is provided,else "first"
def format_name(first,last=""):
  if last:
    return f"{last} , {first}"
  else:
    return 'first'
format_name(first='podalakur' , last='nellore')
```

```python
#recursive function factorial(n) that returns a fatorial of a number
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
factorial(4)
```

```python
#From ['a b','c d e'], produce a flat list of all words.
words=['a b', 'c d e']
result=list(map(lambda x:x.split(),words))
flat_list= [word  for group in result for word in group]
print(flat_list)
```

```python
#	Given students = [{'name':'A','score':80}, {'name':'B','score':60}], Create a list of names of students with score ≥ 75.
students=[{'name':'A','score':80},{'name':'B','score':60}]
result=list(map(lambda s: s['name'],filter(lambda s: s['score']>=75,students)))
print(result)
```
